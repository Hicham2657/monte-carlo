```python
from mpmath import gammainc
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 100
T = 3
step = T/N
```

```python
# brownian simulation
def generate_brownian(N):
    G = np.random.normal(0, 1, N)
    step = T/N
    increment = np.sqrt(step) * G
    brownian = np.concatenate(([0.0], np.cumsum(increment)))
    return brownian
```

```python
# estimate of expected val of (min (tau_a, T) 

theoretical_val = T *(1 - gammainc(1/2, (a**2)/(2*T))/(np.sqrt(np.pi))) + (a**2 / (2* np.sqrt(np.pi))) * gammainc(-1/2, (a**2)/(2*T))

M = 500000
sum = 0.0
a = 1

for _ in range(M):
    tau_a_j = T
    first_passage = True
    brownian_j = generate_brownian(N)
    for i in range(N+1):
        if brownian_j[i] >= a and first_passage:
            tau_a_j = i*T/N
            first_passage = False
            break

    sum += tau_a_j

estimate = sum/M
print(" Theoretical value: ", theoretical_val)
print("estimate : ", estimate)
print("spread = ", estimate - theoretical_val)
```

```python
def MC_estimate_expected_val(N):
    sum = 0.0
    for _ in range(M):
        tau_a_j = T
        first_passage = True
        brownian_j = generate_brownian(N)
        for i in range(N+1):
            if brownian_j[i] >= a and first_passage:
                tau_a_j = i*T/N
                first_passage = False
                break

        sum += tau_a_j

    estimate = sum/M
    return estimate
MC_estimate_expected_val(1000)
```

```python
# spread function of N

def compute_spread(N):
    estimate = MC_estimate_expected_val(N)
    return estimate - theoretical_val

compute_spread(100)
```

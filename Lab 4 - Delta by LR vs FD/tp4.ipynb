```python
import numpy as np
from scipy.stats import norm
S0=K=100.0
T=2
r=0.02
sigma = 0.25
M = 10000
```

```python
def payoff(spot):
    return np.where(spot <= 1.3 * K, np.maximum(spot - K, 0.0), 0.0)
```

```python
def call_bs(spot):
    d1 = (np.log(spot/K) + (r+sigma*sigma*0.5)*T)/(sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    return spot*norm.cdf(d1) - K*norm.cdf(d2)*np.exp(-r*T)
```

```python
def call_payoff(spot):
    spot = np.asarray(spot, dtype=float)
    return np.maximum(spot-K, 0.0)
call_payoff([100, 120])
```

```python
def delta_bs(spot):
    return norm.cdf((np.log(spot/K) + (r+sigma*sigma*0.5)*T)/(sigma*np.sqrt(T)))
delta_bs(S0)
```

```python
def finite_difference_delta(epsilon):
    G = np.random.normal(0,1,M)
    expo = np.exp((r - (sigma*sigma)/2 )* T + sigma*np.sqrt(T)*G)
    up  = payoff((S0*(1+epsilon))* expo)
    down = payoff((S0*(1-epsilon))* expo)
    return np.exp(-r * T) * np.mean(up-down) / (2*epsilon*S0)
```

```python
epss = [0.001, 0.01, 0.1, 1, 10]
for eps in epss:
    print(f'for {eps} : {finite_difference_delta(eps)}')
print(f'for {S0*M**(-1/5)} : { finite_difference_delta(M**(-1/5))}')
```

```python
def likelihood_ratio_delta():
    G = np.random.normal(0,1,M)
    X_theta = S0*np.exp((r - (sigma*sigma)/2 )* T + sigma*np.sqrt(T)*G)
    denom = sigma*T*S0
    W_T = np.sqrt(T)*G
    return np.exp(-r * T) * np.mean(payoff(X_theta)*W_T)/denom
likelihood_ratio_delta()
```
